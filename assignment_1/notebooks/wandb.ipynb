```python
import sys
import os
import numpy as np
import wandb

# 1. Add the src directory to the system path so we can import our data loader
# Assuming your notebook is inside the notebooks/ folder
sys.path.append(os.path.abspath('../src'))
from utils.data_loader import load_and_preprocess_data

# 2. Initialize a new W&B run specifically for this data exploration task
wandb.init(project="da6401_assignment_1", name="q2_1_data_exploration", job_type="viz")

# 3. Load the MNIST dataset
print("Loading data...")
X_train, _, y_train, _ = load_and_preprocess_data('mnist')

# y_train is one-hot encoded (e.g., [0, 0, 1, 0...]). 
# We use np.argmax to convert it back to simple class integers (0-9)
y_class_indices = np.argmax(y_train, axis=1)

# 4. Define the structure of our W&B Table
columns = ["Class Label", "Image 1", "Image 2", "Image 3", "Image 4", "Image 5"]
table = wandb.Table(columns=columns)

# 5. Loop through exactly 10 classes (0 through 9)
for class_id in range(10):
    # Find all indices in the dataset where the label matches current class_id
    indices = np.where(y_class_indices == class_id)[0]
    
    # Grab the first 5 indices for this class
    sample_indices = indices[:5]
    
    # Start building the row with the class label
    row = [str(class_id)]
    
    for idx in sample_indices:
        # Get the flattened image (784,) and reshape it to (28, 28) for visualization
        img_flat = X_train[idx]
        img_2d = img_flat.reshape(28, 28)
        
        # Convert the numpy array to a W&B Image object
        wandb_img = wandb.Image(img_2d)
        row.append(wandb_img)
        
    # Add the completed row to the table
    table.add_data(*row)

# 6. Log the table to the W&B dashboard
print("Logging table to W&B...")
wandb.log({"MNIST_Class_Samples": table})

# 7. Close the run
wandb.finish()
print("Done! Check your W&B dashboard.")
```
